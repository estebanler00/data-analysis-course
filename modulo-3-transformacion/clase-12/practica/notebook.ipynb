`np.where()` permite localizar las posiciones donde se cumple una condición, es el equivalente de un filtro, pero sobre un array, devolviendo los índices en lugar de los valores.

**Consigna:** Responde ¿Cuántas temperaturas están por encima de 30°C?, ¿En qué posiciones (índices) están esas temperaturas? Usá np.where(). Indica ¿Cuál es el promedio de las temperaturas que superan 30°C? ¿Y el promedio de las que están por debajo de 15°C?

Pista: np.where(condicion) devuelve una tupla, el primer elemento es el array de índices. Puedes usarlo para indexar `temperaturas[np.where(temperaturas > 30)]`

```python
import numpy as np

np.random.seed(7)
temperaturas = np.random.normal(22, 5, 100).round(1)

mas30 = np.where(temperaturas >30)
menos15= np.where(temperaturas <15)

promedio_mas30 = np.mean(temperaturas[mas30])
promedio_menos15 = np.mean(temperaturas[menos15])

print(f"Promedio de temperaturas > 30: {promedio_mas30.round(2)}")
print(f"Promedio de temperaturas < 15: {promedio_menos15.round(2)}")
```

### Ejercicio 5 — Calcular correlación con NumPy

El artículo mostró cómo calcular la correlación de Pearson desde cero usando la fórmula. Este ejercicio te pide implementar esa misma lógica en una función reutilizable.

**Consigna:** Escribe una función `correlacion_pearson(x, y)` que calcule la correlación usando únicamente NumPy — sin usar `np.corrcoef()` ni `.corr()` de Pandas. Después verifica que tu función da el mismo resultado que ambas para el par `horas_estudio` y `puntaje_final`.

```python
import numpy as np
import pandas as pd

df = pd.read_csv("estudiantes.csv")

def correlacion_pearson(x, y):
    meanX = np.mean(x)
    meanY = np.mean(y)
    varX = x - meanX
    varY = y - meanY
    numerator = np.sum(varX * varY)
    denominator = np.sqrt(np.sum(varX ** 2) * np.sum(varY **2))
    correlacion = numerator / denominator
    return correlacion

x = df["horas_estudio"].values
y = df["puntaje_final"].values

r_manual = correlacion_pearson(x, y)
r_numpy  = np.corrcoef(x, y)[0, 1]
r_pandas = df["horas_estudio"].corr(df["puntaje_final"])

print(f"Manual:  {r_manual:.6f}")
print(f"NumPy:   {r_numpy:.6f}")
print(f"Pandas:  {r_pandas:.6f}")

# ¿Los tres dan el mismo resultado?
```

### Ejercicio 10 — Escanear todo el dataset

En un análisis real no revisás columna por columna manualmente — escribís una función que escanee todo el dataset y produzca un reporte de outliers por columna.

**Consigna:** Escribe una función `reporte_outliers(df, metodo="iqr", umbral=3)` que:

1. Recorra todas las columnas numéricas del DataFrame
2. Aplique el método indicado (`"iqr"` o `"zscore"`)
3. Devuelva un DataFrame resumen con: columna, cantidad de outliers, porcentaje sobre el total, y los valores detectados

```python
import numpy as np
import pandas as pd

df = pd.read_csv("estudiantes.csv")

def reporte_outliers(df, metodo="iqr", umbral=3):
    resultados = []

    for col in df.select_dtypes(include=[np.number]).columns:
        arr = df[col].values

        if metodo == "iqr":
            p25 = np.percentile(arr, 25)
            p75 = np.percentile(arr, 75)
            iqr = p75 - p25
            mask = (arr < p25 - 1.5 * iqr) | (arr > p75 + 1.5 * iqr)
        elif metodo == "zscore":
            zscores = (arr - np.mean(arr)) / np.std(arr)
            mask = np.abs(zscores) > umbral

        valores = arr[mask].tolist()
        resultados.append({
            "columna":    col,
            "outliers":   mask.sum(),
            "porcentaje": round(mask.sum() / len(arr) * 100, 1),
            "valores":    valores,
        })

    return pd.DataFrame(resultados)

print("=== IQR ===")
print(reporte_outliers(df, metodo="iqr"))

print("\n=== Z-score ===")
print(reporte_outliers(df, metodo="zscore"))

# ¿Los cuatro outliers intencionales aparecen en alguno de los dos reportes?
# ¿Hay alguno que solo detecta uno de los dos métodos?
```
