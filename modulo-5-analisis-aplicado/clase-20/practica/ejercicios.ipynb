```python
import pandas as pd

df = pd.read_csv("/Users/esteb/Documents/Esteban/CURSOS/Data-analisis-ComIT_ok/data-analysis-course/modulo-5-analisis-aplicado/clase-20/ventas_kiosco.csv")

df["total_venta"] = df["cantidad"] * df["precio_unitario"]

df["fecha"] = pd.to_datetime(df["fecha"], format="%Y-%m-%d")
df["mes"] = df["fecha"].dt.month

marzo = df[df["mes"] == 3] 
febrero = df[df["mes"] == 2]

total_ventas_marzo = marzo["total_venta"].sum()
total_ventas_febrero = febrero["total_venta"].sum()

print(f" Total ventas Marzo: ${total_ventas_marzo} \n \n Total ventas Febrero: ${total_ventas_febrero}")
```

```python
#Ticket Promedio 

ticket_promedio_marzo = (total_ventas_marzo / len(marzo)).round(2)

ticket_promedio_febrero = (total_ventas_febrero / len(febrero)).round(2)


print(f"Ticket Promedio por Mes: \n Marzo: \n ${ticket_promedio_marzo} \n \n Febrero: \n ${ticket_promedio_febrero}")
```

```python
#Clientes Únicos

cu_febrero = febrero["cliente"].nunique()
cu_marzo = marzo["cliente"].nunique()

print(f"Clientes únicos Marzo: {cu_marzo} \nClientes únicos Febrero: {cu_febrero}")

# O tambien podria ser:
#clientes_unicos_mensual = df.groupby("mes")["cliente"].nunique()
#print(f"\n Cantidad de clientes únicos por mes:\n  {clientes_unicos_mensual}")

vcu_febrero = febrero["cliente"].unique()
vcu_marzo = marzo["cliente"].unique()

set_feb = set(vcu_febrero)
set_mar = set(vcu_marzo)

clientes_unicos_marzo = set_mar - set_feb
clientes_unicos_marzo_sorted = sorted(clientes_unicos_marzo)

print(f"\n Primeros clientes en Marzo: \n Cantidad: {len(clientes_unicos_marzo)} \n Clientes: {clientes_unicos_marzo_sorted}")

compra_por_cliente = df.groupby("cliente").size()

print(f"\n Cantidad de compras por cliente: \n {compra_por_cliente}")
```
