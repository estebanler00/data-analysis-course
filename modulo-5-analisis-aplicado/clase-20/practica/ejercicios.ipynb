```python
import pandas as pd

df = pd.read_csv("/Users/esteb/Documents/Esteban/CURSOS/Data-analisis-ComIT_ok/data-analysis-course/modulo-5-analisis-aplicado/clase-20/ventas_kiosco.csv")

df["total_venta"] = df["cantidad"] * df["precio_unitario"]

df["fecha"] = pd.to_datetime(df["fecha"], format="%Y-%m-%d")
df["mes"] = df["fecha"].dt.month

marzo = df[df["mes"] == 3] 
febrero = df[df["mes"] == 2]

total_ventas_marzo = marzo["total_venta"].sum()
total_ventas_febrero = febrero["total_venta"].sum()

print(f" Total ventas Marzo: ${total_ventas_marzo} \n \n Total ventas Febrero: ${total_ventas_febrero}")
```

```python
#Ticket Promedio 

ticket_promedio_marzo = (total_ventas_marzo / len(marzo)).round(2)

ticket_promedio_febrero = (total_ventas_febrero / len(febrero)).round(2)


print(f"Ticket Promedio por Mes: \n Marzo: \n ${ticket_promedio_marzo} \n \n Febrero: \n ${ticket_promedio_febrero}")
```

```python
#Clientes Únicos

clientes_unicos_marzo = df.groupby(["cliente", "mes"])["cliente"].count()

print(clientes_unicos_marzo)
```
